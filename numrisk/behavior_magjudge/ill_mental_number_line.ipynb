```python
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
import pandas as pd
x = np.arange(0, 20, 0.1)
y1 = stats.norm.pdf(x, 6, 1)
y2 = stats.norm.pdf(x, 12, 1)

# Fill the area under the curves
plt.figure(figsize=(4,2))
plt.fill_between(x, y2, alpha=0.3, color='yellow')
plt.fill_between(x, y1, alpha=0.3, color='blue')

# Plot the curve lines on top
plt.plot(x, y2, c='yellow')
plt.plot(x, y1, c='blue')
plt.ylabel(None)
plt.yticks([])
plt.ylim(0, 0.6)
plt.xlabel('Anzahl')
plt.title('Mentale Representation')
sns.despine()
```

```python
def plot_mental_number_lin_comparison(x,y1,y2):
    # Fill the area under the curves
    plt.figure(figsize=(4,2))
    plt.fill_between(x, y2, alpha=0.3, color='yellow')
    plt.fill_between(x, y1, alpha=0.3, color='blue')

    # Plot the curve lines on top
    plt.plot(x, y2, c='yellow')
    plt.plot(x, y1, c='blue')
    plt.ylabel(None)
    plt.yticks([])
    plt.ylim(0, 0.6)
    plt.xlabel('Anzahl')
    plt.title('Mentale Representation')
    sns.despine()
```

```python
x = np.arange(0, 20, 0.1)
y1 = stats.norm.pdf(x, 6, 1) # blue
y2 = stats.norm.pdf(x, 12, 1) # yellow
plot_mental_number_lin_comparison(x,y1,y2)

y1 = stats.norm.pdf(x, 12, 1)
y2 = stats.norm.pdf(x, 9, 1)

plot_mental_number_lin_comparison(x,y1,y2)

```
